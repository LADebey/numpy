### EX 03 SLICING

```python
import numpy as np

array1= np.arange(1,101)

odd = array1[::2]

evenReverse = array1[1::2][::-1]

array3Final = array1.copy()
array3Final[1::3] = 0


print( "1 To 100", array1)
print ("impair", odd)
print("even reverse", evenReverse)
print ("Jump 3 and add a zero", array3Final)
```

### EX 04 RANDOM

### EX 05 SPLIT CONCAT RESHAPE

```python
import numpy as np 

firstArr = np.arange(1, 51)

secArr = np.arange(51, 101)

newArr = np.concatenate([firstArr,secArr])

Matrix = newArr.reshape(10,10)

print(firstArr)
print(secArr)
print(newArr)

print(Matrix)
```
